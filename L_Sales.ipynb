<h2>⚖️ Step 4: Pros &amp; Cons of the Improved Model</h2>

<div class="pros-cons">
    <div class="pros">
        <h4 style="color:#0d9488;">✅ Pros</h4>
        <ul>
            <li><strong>Massive fit improvement:</strong> R² jumps from 31% → 97%.</li>
            <li><strong>Quantifies the disruption:</strong> Week 3 cost ~18.8 bags.</li>
            <li><strong>Reveals true trend:</strong> Underlying growth is +2.07 bags/week (not +2.36).</li>
            <li><strong>Easy to explain:</strong> Stakeholders understand a "one-off flag."</li>
        </ul>
    </div>
    <div class="cons">
        <h4 style="color:#dc2626;">❌ Cons</h4>
        <ul>
            <li><strong>Overfitting risk:</strong> Only 7 data points with 3 variables.</li>
            <li><strong>Poor for forecasting:</strong> Week 8 forecast is same as original model (D3=0).</li>
            <li><strong>No root cause:</strong> It flags <em>that</em> Week 3 was bad, but not <em>why</em>.</li>
            <li><strong>Fails on new disruptions:</strong> If Week 8 drops, the model breaks.</li>
        </ul>
    </div>

<h2>🎯 Step 5: Final Verdict &amp; Recommendation</h2>

<div class="box box-purple">
    <h3 style="margin-top:0;">Use the Improved Model for Explanatory Analysis</h3>
    <p>
        Use the <strong>D3 model (R² = 97%)</strong> to explain to leadership that <strong>Week 3 was a one-off disruption</strong>,
        and that the underlying trend is actually a healthy <strong>+2.07 bags per week</strong>.
    </p>
    <hr style="border-color:#cbd5e1; margin:20px 0;">
    <h3 style="margin-top:0;">Do NOT use it for Forecasting Week 8</h3>
    <p>
        Because <code>D3 = 0</code> for Week 8, the forecast is identical to the original model.
        Instead, use a <strong>3-week moving average</strong> or the slope anchored to Week 7:
    </p>
    <div style="background:#eef2ff; padding:12px 18px; border-radius:8px; display:inline-block;">
        Forecast Week 8 ≈ 54 + 2.07 = <strong>56.1 bags</strong>
    </div>
    <p style="margin-top:15px;">
        Wait until you have <strong>at least 20 weeks</strong> of data before using dummy variables for official forecasting.
    </p>
</div>
